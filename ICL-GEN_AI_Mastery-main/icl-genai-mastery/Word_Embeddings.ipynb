```python
from sklearn.preprocessing import OneHotEncoder

words = [
    ['python'],
    ['java'],
    ['machine'],
    ['learning'],
    ['python'],
    ['ai']
]

encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)

print("Words:", encoder.categories_[0])
print("Embeddings:")
print(result)
```

```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["python", "is", "easy"],
    ["python", "is", "powerful"],
    ["machine", "learning", "uses", "data"],
    ["deep", "learning", "uses", "neural", "networks"],
    ["python", "is", "used", "for", "machine", "learning"],
    ["deep", "learning", "is", "part", "of", "ai"]
]
```

```python
model = Word2Vec(
    sentences,
    vector_size=10,
    window=2,
    min_count=1,
    workers=1
)
```

```python
print(model.wv.most_similar('python', topn=3))
```

```python
print(model.wv.most_similar('learning', topn=3))
```
