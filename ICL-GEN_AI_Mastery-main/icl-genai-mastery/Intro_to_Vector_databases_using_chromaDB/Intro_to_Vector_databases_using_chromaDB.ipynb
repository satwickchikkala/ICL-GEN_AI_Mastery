```python
import chromadb

client = chromadb.Client()

print("ChromaDB client created successfully!")
```

```python
collection = client.create_collection(
    name="genai_rag_collection"
)

print("Collection created:", collection.name)
```

```python
collection.add(
    documents=[
        "Generative AI can create new content such as text, images, audio, video, and code.",
        "Large Language Models are trained on large amounts of text data.",
        "Transformers use attention mechanisms to understand relationships between words.",
        "Prompt engineering helps users create effective instructions for AI models.",
        "RAG retrieves relevant information from external documents before generating an answer.",
        "Document chunking divides large documents into smaller pieces for efficient retrieval.",
        "Embeddings represent text as numerical vectors that capture semantic meaning.",
        "Vector databases store embeddings and support similarity search.",
        "ChromaDB is a vector database that can be used to store and retrieve embeddings.",
        "A RAG system combines document retrieval with a language model."
    ],
    ids=[
        "genai1",
        "genai2",
        "genai3",
        "genai4",
        "genai5",
        "genai6",
        "genai7",
        "genai8",
        "genai9",
        "genai10"
    ]
)

print("Documents added!")
print("Total documents:", collection.count())
```

Querying_Our_Docs

```python
results = collection.query(
    query_texts=["How does RAG retrieve information?"],
    n_results=3
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
collection.add(
    documents=[
        "Prompt engineering involves designing clear instructions for AI models.",
        "ChromaDB can store embeddings and perform similarity searches.",
        "RAG systems retrieve relevant documents before generating responses."
    ],

    metadatas=[
        {"category": "prompt_engineering"},
        {"category": "vector_database"},
        {"category": "rag"}
    ],

    ids=[
        "extra1",
        "extra2",
        "extra3"
    ]
)

print("Additional documents added!")
```

```python
results = collection.query(
    query_texts=["How does retrieval augmented generation work?"],
    n_results=2,
    where={"category": "rag"}
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")
```

Using_Sentence_Transformers

```python
from chromadb.utils import embedding_functions

hf_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)
```

```python
custom_collection = client.create_collection(
    name="genai_custom_embeddings",
    embedding_function=hf_embedder
)
```

```python
custom_collection.add(
    documents=[
        "Generative AI creates new content using learned patterns.",
        "Large language models understand and generate human-like text.",
        "RAG retrieves external information to improve AI responses.",
        "Vector databases enable fast similarity search over embeddings."
    ],

    ids=[
        "custom1",
        "custom2",
        "custom3",
        "custom4"
    ]
)
```

```python
results = custom_collection.query(
    query_texts=["How can AI retrieve information from documents?"],
    n_results=2
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")
```
