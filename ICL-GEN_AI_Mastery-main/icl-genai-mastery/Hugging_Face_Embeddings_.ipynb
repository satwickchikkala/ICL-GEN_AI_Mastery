```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "Python is widely used for artificial intelligence and machine learning.",
    "Machine learning allows computers to learn patterns from data.",
    "Deep learning uses neural networks with multiple layers.",
    "Natural language processing helps computers understand human language.",
    "Computer vision allows machines to understand images and videos.",
    "SQL is used to store and retrieve data from relational databases.",
    "React is a JavaScript library used to build user interfaces.",
    "Docker is used to package applications into containers.",
    "Cybersecurity protects systems and networks from attacks.",
    "Cloud computing provides computing resources over the internet."
]
```

```python
doc_embeddings = model.encode(documents)
```

```python
def semantic_search(query, top_k=3):
    query_embedding = model.encode([query])

    scores = cosine_similarity(
        query_embedding,
        doc_embeddings
    )[0]

    ranked_idx = scores.argsort()[::-1][:top_k]

    print(f"Query: {query}\n")

    for idx in ranked_idx:
        print(f"Score: {scores[idx]:.3f} | {documents[idx]}")
```

```python
semantic_search("How are computers able to learn from data?")
```

```python
semantic_search("How can I protect my computer from hackers?")
```

```python
model2 = SentenceTransformer("all-mpnet-base-v2")

doc_embeddings2 = model2.encode(documents)

query = "How do computers learn from information?"

query_embedding2 = model2.encode([query])

scores2 = cosine_similarity(
    query_embedding2,
    doc_embeddings2
)[0]

ranked_idx = scores2.argsort()[::-1][:3]

for idx in ranked_idx:
    print(f"{scores2[idx]:.3f} | {documents[idx]}")
```

```python
faq_questions = [
    "What is Python?",
    "What is machine learning?",
    "What is deep learning?",
    "What is SQL?",
    "What is React?"
]

faq_answers = [
    "Python is a popular programming language used for AI, data science, automation, and web development.",
    "Machine learning is a technique where computers learn patterns from data.",
    "Deep learning is a type of machine learning that uses neural networks with multiple layers.",
    "SQL is a language used to store, manage, and retrieve data from relational databases.",
    "React is a JavaScript library used to build user interfaces."
]
```

```python
faq_embeddings = model.encode(faq_questions)
```

```python
def faq_bot(query):

    query_embedding = model.encode([query])

    scores = cosine_similarity(
        query_embedding,
        faq_embeddings
    )[0]

    best_index = scores.argmax()

    print("Question:", query)
    print("Answer:", faq_answers[best_index])
    print("Similarity:", round(scores[best_index], 3))
```

```python
faq_bot("Can you explain what Python is?")
```

```python
faq_bot("How does a computer learn from data?")
```
